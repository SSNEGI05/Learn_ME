Section 1.1: Object-Oriented Programming (OOP)
The Problem: "Analyst Spaghetti Code"
As an analyst, we often write "procedural" code. If we want to calculate the yield for 10 different bonds, we might create 10 different variables or a giant nested dictionary.

The Risk: If you change how the yield is calculated, you have to find and change it in 10 different places.

The Mess: Data (price, ticker) and Logic (math formulas) are scattered everywhere.

The Solution: Object-Oriented Programming (OOP)
OOP allows us to "bundle" Data and Logic together into a single unit called a Class. Instead of thinking about "lines of code," we start thinking about "Entities" (like a Stock, a Model, or a Portfolio).

##🏛️ Topic 1: What is a Class?

The Logic (The "Why"):
Think of a Class as a Security Master Template. Every time a company goes public, they don't reinvent the concept of a "Stock." They follow a pre-defined blueprint: it must have a ticker, a listing exchange, and a way to pay dividends.

The Concept (The "How"):

Class (The Blueprint): This is the code that defines the structure. It doesn't hold real data yet; it just says what data will be there.

Object/Instance (The Real Asset): This is when you take the blueprint and fill it with real information (e.g., "NVDA", $800, "Tech").

Attributes: These are the characteristics (The "Nouns").

Methods: These are the functions inside the class (The "Verbs").

```python
class FinancialInstrument:
    """
    A base class representing a generic financial instrument.
    This is our 'Blueprint'.
    """

    def __init__(self, ticker: str, price: float, sector: str):
        """
        The __init__ method runs automatically when we create a new object.
        'self' ensures the data stays with the specific object (e.g., AAPL's price vs TSLA's).
        """
        # Attributes (Data)
        self.ticker = ticker
        self.price = price
        self.sector = sector

    def display_info(self):
        """
        A 'Method' (Action) that our object can perform.
        """
        print(f"--- Asset Report ---")
        print(f"Ticker: {self.ticker}")
        print(f"Price:  ${self.price}")
        print(f"Sector: {self.sector}\n")

# --- CREATING OBJECTS (Instances) ---
# We use the blueprint to create specific 'Objects'
asset_1 = FinancialInstrument("AAPL", 185.20, "Technology")
asset_2 = FinancialInstrument("GLD", 2150.50, "Commodities")

# Let's test them
asset_1.display_info()
asset_2.display_info()
```

##🏛️ Topic 2: Understanding self and Attribute Scope
**1. The self Keyword (The "Which one?" Pointer)**
In Python, self is not a keyword by technicality (you could call it banana), but by convention, it is always named self.

The Logic: Imagine you have two bank accounts (Savings and Checking). When you call the method withdraw(), Python needs to know which account to subtract money from.

The Concept: self is a placeholder that says: "Apply this action to the specific object that called me."

**2. Instance Attributes vs. Class Attributes**
This is a critical distinction for "Type 2" engineering.

Instance Attributes (self.variable): Unique to each object. (e.g., Every stock has its own price).

Class Attributes (Shared): Data that is the same for every single object created from that class. (e.g., Every stock in your system might be traded in the same currency).


```python
class SmartPortfolio:
    # --- CLASS ATTRIBUTES (Shared by everyone) ---
    exchange = "NSE"
    currency = "INR"

    def __init__(self, owner_name: str):
        # --- INSTANCE ATTRIBUTES (Unique to each user) ---
        self.owner = owner_name
        self.holdings = []  # List to store tickers

    def add_stock(self, ticker: str):
        """Action: Add a stock to THIS specific portfolio"""
        self.holdings.append(ticker)
        print(f"Added {ticker} to {self.owner}'s portfolio.")

# --- TESTING THE SCOPE ---

# Create two separate portfolios
my_portfolio = SmartPortfolio("Shubham")
client_portfolio = SmartPortfolio("Retail_Client")

# 1. Instance Attributes are DIFFERENT
my_portfolio.add_stock("RELIANCE")
print(f"Shubham's Holdings: {my_portfolio.holdings}")
print(f"Client's Holdings: {client_portfolio.holdings}") # Will be empty!

# 2. Class Attributes are the SAME
print(f"Shubham's Exchange: {my_portfolio.exchange}")
print(f"Client's Exchange: {client_portfolio.exchange}")

# 3. Changing a Class Attribute changes it for EVERYONE
SmartPortfolio.exchange = "BSE"
print(f"Updated Exchange for all: {my_portfolio.exchange}")
```

```python
class OptionContract:
    # --- CLASS ATTRIBUTE ---
    # This is true for EVERY contract created under this class
    underlying_market = "Equity Derivatives"

    def __init__(self, ticker: str, strike: float):
        # --- INSTANCE ATTRIBUTES ---
        # These are unique to each specific trade
        self.ticker = ticker
        self.strike = strike

    def print_details(self):
        # Without 'self', Python wouldn't know WHICH ticker to print
        print(f"Contract: {self.ticker} | Strike: {self.strike} | Market: {self.underlying_market}")

# --- THE "AHA!" MOMENT ---

trade_1 = OptionContract("RELIANCE", 3000)
trade_2 = OptionContract("INFY", 1600)

# 1. Accessing unique data
trade_1.print_details() # Knows it's RELIANCE because of 'self'
trade_2.print_details() # Knows it's INFY because of 'self'

# 2. What happens if we change the CLASS attribute?
print("\n--- Changing Market Rule to 'Global' ---")
OptionContract.underlying_market = "Commodity"

trade_1.print_details() # Now says 'Commodity'
trade_2.print_details() # Also says 'Commodity' because they SHARE this variable.
```

##🏛️ Topic 3: Methods — Making the Blueprint "Do" Things
1. What is a Method?
A method is just a function that lives inside a class.

The Logic: You don't want a "naked" function floating around your code called calculate_var(). You want that function to be part of a Portfolio class so it automatically knows which assets to look at.

The Concept: Methods use self to access the attributes (data) we stored earlier.

2. The Power of "Encapsulation"
This is a fancy "Type 2" word. It means we hide the complex math inside the method.

User Experience: The person using your code just types my_trade.get_profit().

The Reality: Inside that method, you might have 50 lines of complex Greek calculations. The user doesn't need to see the "mess"; they just need the result.

```python
class Trade:
    def __init__(self, ticker: str, entry_price: float, quantity: int):
        self.ticker = ticker
        self.entry_price = entry_price
        self.quantity = quantity
        self.status = "OPEN" # Every trade starts as Open

    # --- THIS IS A METHOD (The Verb) ---
    def calculate_pnl(self, current_market_price: float) -> float:
        """
        Calculates the profit or loss based on a new market price.
        Notice how it uses 'self' to find the original entry_price.
        """
        pnl = (current_market_price - self.entry_price) * self.quantity
        return pnl

    def close_trade(self, exit_price: float):
        """Changes the internal state of the object"""
        final_pnl = self.calculate_pnl(exit_price)
        self.status = "CLOSED"
        print(f"✅ Trade on {self.ticker} CLOSED. Final P&L: {final_pnl:,.2f}")

# --- EXECUTION ---
my_trade = Trade("NVDA", 800, 10) # Bought 10 shares at 800

# Let's check the P&L if price moves to 850
current_pnl = my_trade.calculate_pnl(850)
print(f"Current Unrealized P&L for {my_trade.ticker}: {current_pnl}")

# Now let's close it at 900
my_trade.close_trade(900)
```

##🏛️ Topic 4: Inheritance — The "Parent-Child" Relationship
1. The Logic (The "Why")
In ML Engineering, you don't want to write a completely new script for every different model. You create a BaseModel that handles data loading and error checking, and then you "Inherit" from it to create a RegressionModel or a TransformerModel.

2. The Concept (The "How")
Parent Class: The master template containing the "Common" code.

Child Class: The specialized template that takes everything from the Parent and adds its own "Secret Sauce."

The **super()** function: This is a special command that lets the Child call a function from the Parent (like asking your parents for the family recipe before adding your own spices).

```python
# --- THE PARENT CLASS (General) ---
class Security:
    def __init__(self, ticker: str, price: float):
        self.ticker = ticker
        self.price = price

    def describe(self):
        print(f"Asset: {self.ticker} | Current Price: {self.price}")

# --- THE CHILD CLASS 1 (Specialized for Stocks) ---
# Notice the (Security) in the brackets—that means it INHERITS from Security
class Stock(Security):
    def __init__(self, ticker: str, price: float, dividend: float):
        # super() calls the Parent's __init__ to handle ticker and price
        super().__init__(ticker, price)
        self.dividend = dividend # Unique to Stocks

    def calculate_yield(self):
        return (self.dividend / self.price) * 100

# --- THE CHILD CLASS 2 (Specialized for Bonds) ---
class Bond(Security):
    def __init__(self, ticker: str, price: float, coupon: float):
        super().__init__(ticker, price)
        self.coupon = coupon # Unique to Bonds

    def describe(self):
        # We can OVERRIDE the parent's method to be more specific
        print(f"FIXED INCOME: {self.ticker} | Coupon: {self.coupon}% | Price: {self.price}")

# --- EXECUTION ---
my_stock = Stock("ITC", 450, 15.5)
my_bond = Bond("GS2030", 98.5, 7.2)

# Both can use methods from the Parent
my_stock.describe()
print(f"Dividend Yield: {my_stock.calculate_yield():.2f}%")

# The Bond uses its OWN version of 'describe'
my_bond.describe()
```

🏛️ Topic 6: Dunder Methods (Magic Methods)
"Dunder" stands for Double Underscore (e.g., __init__). These are special methods that Python calls automatically behind the scenes.

1. The Logic (The "Why")
Have you ever wondered why you can use + to add two numbers, but you can't use + to add two "Stock" objects? Dunder methods allow you to "teach" Python how to handle your custom objects using standard symbols like +, -, or even printing.

2. Common Dunder Methods
__init__: The Constructor. Sets up the object.

__str__ or __repr__: Controls what happens when you print(object). Instead of seeing <__main__.Stock object at 0x123>, you see "Stock: AAPL".

__len__: Controls what happens when you call len(object).

__call__: This is the "Magic" one for ML. It allows you to treat an object like a function. (This is exactly how PyTorch models work).

```python
class SmartStock:
    def __init__(self, ticker: str, price: float):
        self.ticker = ticker
        self.price = price

    # 1. MAGIC METHOD: The String Representation
    def __str__(self):
        return f"📈 [SmartStock] {self.ticker} is trading at ${self.price}"

    # 2. MAGIC METHOD: The Function Caller
    def __call__(self, quantity: int):
        """
        Allows us to do: my_stock(10) to get the total value.
        In ML, this is used to 'run' a model: my_model(input_data).
        """
        total_value = self.price * quantity
        print(f"💰 Total value for {quantity} shares of {self.ticker}: ${total_value:,.2f}")
        return total_value

# --- EXECUTION ---
apple = SmartStock("AAPL", 185.00)

# This triggers __str__ automatically
print(apple)

# This triggers __call__ automatically
# We treat the OBJECT like a FUNCTION!
apple(50)
```

Section 1.2: Generators & Iterators — Handle Huge Datasets Efficiently
The Problem: The "Memory Wall"
As an analyst, you might be used to loading a CSV into a Pandas DataFrame.

The Risk: If a CSV is 10GB and your RAM is 8GB, your computer crashes.

The Inefficiency: Even if it fits, why load 1 million rows if you only need to look at them one by one to check for a signal?

The Solution: "Lazy Evaluation"
Instead of a "List" (which is like a warehouse full of goods), we use Generators (which are like a conveyor belt).

Iterators: Objects that allow you to go through a collection one by one using the next() function.

Generators: A special type of function that "yields" data instead of "returning" it. It pauses its execution and remembers where it left off.

🏛️ Topic 1: The yield Keyword
In a normal function, return kills the function. In a generator, yield says: "Here is your data, now I'm going to take a nap. Wake me up when you want the next piece."

```python
import sys
import random
import time

# 1. THE TRADITIONAL WAY (List)
def get_prices_list(n):
    prices = []
    for i in range(n):
        prices.append(random.uniform(100, 200))
    return prices

# 2. THE TYPE 2 WAY (Generator)
def get_prices_generator(n):
    for i in range(n):
        # We 'yield' the price instead of building a giant list
        yield random.uniform(100, 200)

# --- COMPARISON ---
n = 1000000 # 1 Million records

# Check memory of a List
price_list = get_prices_list(n)
print(f"📦 List Memory Size: {sys.getsizeof(price_list) / (1024*1024):.2f} MB")

# Check memory of a Generator
price_gen = get_prices_generator(n)
print(f"⚙️ Generator Memory Size: {sys.getsizeof(price_gen) / 1024:.2f} KB")

print("\n--- Fetching first 3 prices from Generator ---")
print(next(price_gen))
print(next(price_gen))
print(next(price_gen))
```

##🏛️ Topic 2: Generator Expressions — The "One-Liner" Efficiency
1. The Logic (The "Why")
Sometimes you don't need a full function with a yield statement. You just want a quick, one-line way to process data without creating a massive list in memory.

The Analyst Way: [x for x in data if x > 100] (This creates a List, taking up RAM).

The Type 2 Way: (x for x in data if x > 100) (This creates a Generator, taking up almost zero RAM).

2. The Concept (The "How")
The only difference in syntax is the brackets:

Square Brackets [] = List Comprehension (Builds the whole "Warehouse" immediately).

Parentheses () = Generator Expression (Sets up the "Conveyor Belt").

```python
import sys

# Suppose we have 1 million stock prices
prices = [100.5, 99.2, 101.8, 105.4] * 250000

# 1. LIST COMPREHENSION (Eager - creates the list now)
# We want to find all prices above 102
high_prices_list = [p for p in prices if p > 102]

# 2. GENERATOR EXPRESSION (Lazy - waits until you ask)
high_prices_gen = (p for p in prices if p > 102)

# --- THE MEMORY TEST ---
print(f"📊 List Size: {sys.getsizeof(high_prices_list) / (1024*1024):.2f} MB")
print(f"⚙️ Generator Size: {sys.getsizeof(high_prices_gen)} bytes")

# --- THE SPEED TEST ---
print("\nFirst 3 high prices from the generator:")
print(next(high_prices_gen))
print(next(high_prices_gen))
print(next(high_prices_gen))
```

Section 1.3: Decorators — Modify Functions Without Changing Them
The Problem: "The Cluttered Analyst"
Imagine you have 20 different trading strategies. You want to measure how long each one takes to run (latency).

The Amateur Way: You add start_time = time.time() and print(end_time - start_time) inside every single function.

The Risk: If you want to change the logging format, you have to manually edit 20 different functions. It's a maintenance nightmare.

The Solution: The "Wrapper"
A Decorator is a function that "wraps" another function. It allows you to execute code before and after the main function runs, without ever touching the original code.

📊 The Logic: The "Shipping Box" Analogy
Think of your core trading logic as a Product (like a smartphone). The Decorator is the Shipping Box.

The box protects the product (Security).

The box has a tracking label (Logging/Timing).

The product inside stays exactly the same.

##🏛️ Topic 1: The @decorator Syntax
In Python, we use the @ symbol above a function to "wrap" it. This is heavily used in PyTorch and FastAPI

```python
import time
import functools

# 1. THE DECORATOR (The Wrapper)
def monitor_performance(func):
    """
    A decorator that measures the execution time of a function.
    @functools.wraps(func) is a 'Type 2' best practice—it keeps
    the original function's name and metadata intact.
    """
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        print(f"🚀 Starting execution of: {func.__name__}")
        start_time = time.perf_counter()

        # Execute the original function
        result = func(*args, **kwargs)

        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"✅ Finished {func.__name__} in {duration:.6f} seconds\n")
        return result
    return wrapper

# 2. APPLYING THE DECORATOR
@monitor_performance
def calculate_complex_alpha(data_points: int):
    """Simulates a heavy quantitative calculation"""
    total = 0
    for i in range(data_points):
        total += i ** 2
    return total

@monitor_performance
def get_market_status(ticker: str):
    """Simulates a quick API check"""
    return f"{ticker} is currently OPEN"

# --- EXECUTION ---
# Notice we just call the functions normally.
# The 'timing' logic happens automatically!
alpha_val = calculate_complex_alpha(1000000)
status = get_market_status("RELIANCE")
```

##🏛️ Topic 2: Financial Use Case — The Auto-Logger
The Logic (The "Audit Trail")
In a professional trading firm (like where you work as a Senior Investment Analyst), every action must be audited. If a model places a trade, we need to know:

Who (Which model/function) called the trade?

What parameters (Ticker, Price, Side) were used?

Result was it successful?

The Concept: The @log_trade Decorator
We will create a decorator that intercepts the arguments of a function (like ticker and quantity), prints them to the console (simulating a file log), and then lets the trade proceed.

```python
import functools
from datetime import datetime

# 1. THE AUDIT DECORATOR
def log_trade(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        # Capture the time and function name
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

        # Log the 'input' data before the trade happens
        print(f"--- [AUDIT LOG | {timestamp}] ---")
        print(f"Action: {func.__name__.upper()}")
        print(f"Parameters: {args} {kwargs}")

        # Execute the actual trade function
        result = func(*args, **kwargs)

        print(f"Status: COMPLETED\n")
        return result
    return wrapper

# 2. APPLYING TO DIFFERENT TRADING FUNCTIONS
@log_trade
def execute_market_buy(ticker: str, qty: int):
    """Simulates hitting the 'Buy' button on Zerodha/Fyers"""
    return f"Order placed for {qty} shares of {ticker}"

@log_trade
def set_stop_loss(ticker: str, price: float):
    """Simulates setting a SL trigger"""
    return f"SL set at {price}"

# --- EXECUTION ---
# You just write your strategy logic; the logging is "invisible"
execute_market_buy("RELIANCE", qty=50)
set_stop_loss("RELIANCE", price=2850.50)
```

Section 1.4: Context Managers — Control Setup and Teardown
The Problem: "The Leaky Analyst"
As an analyst, you might open a file to write results and then forget to close it, or your code might hit an error halfway through.

The Risk: If you don't close a file or a database connection, you "leak" memory. Eventually, your system will run out of resources and crash.

The Mess: You end up writing try...finally blocks everywhere to manually close things, making your code hard to read.

The Solution: The with Statement
A Context Manager is a Python object that knows exactly what to do when you enter a block of code and exactly what to do when you leave it.

Setup: Automatically opens the file/connection.

Teardown: Automatically closes it—even if an error occurs inside the block.

📊 The Logic: The "Borrowed Book" Analogy
Think of a Context Manager like a Library Card.

Entry: You check out a book (The resource is now yours to use).

Action: You read the book (You run your code).

Exit: You return the book (The resource is released for others).
The library doesn't care if you finished the book or fell asleep; the rule is that it must be returned when you leave.

##🏛️ Topic 1: The standard with open()
You have likely seen this before: with open('file.txt', 'r') as f:. This is a built-in Context Manager. But as a Type 2 Engineer, you will often need to build your own custom ones for things like database connections or timing specific blocks of code.

```python
from contextlib import contextmanager

# 1. THE 'TYPE 2' WAY: Using a Decorator to create a Context Manager
@contextmanager
def market_data_connection(server_name: str):
    """
    Simulates a professional connection to a data server.
    """
    print(f"🔌 [SETUP] Connecting to {server_name}...")
    connection_active = True

    try:
        # Everything before the 'yield' is the SETUP
        yield f"DATA_STREAM_{server_name}"

    finally:
        # Everything after 'yield' is the TEARDOWN
        # This code runs NO MATTER WHAT (even if the middle code crashes)
        connection_active = False
        print(f"🔌 [TEARDOWN] Connection to {server_name} safely CLOSED.")

# --- EXECUTION ---

# Scenario A: Everything goes right
print("--- Scenario A: Normal Operation ---")
with market_data_connection("NSE_TICK_SERVER") as stream:
    print(f"📊 Currently reading from: {stream}")
    print("✅ Logic complete.")

print("\n" + "="*40 + "\n")

# Scenario B: The code crashes in the middle
print("--- Scenario B: Code Crashes ---")
try:
    with market_data_connection("BSE_TICK_SERVER") as stream:
        print(f"📊 Currently reading from: {stream}")
        print("💥 Oops! A 'ZeroDivisionError' is about to happen...")
        1 / 0  # This will crash the code
except ZeroDivisionError:
    print("⚠️ Caught the error in the main script.")

# Notice: In Scenario B, the [TEARDOWN] still ran!
```

🏗️ 1. Decorators: The "Standard Operating Procedure" (SOP)
Think of a Decorator as a Compliance Checklist or an SOP that you wrap around a task.

The Analyst Scenario:
You have a function called execute_trade().

The Problem: You shouldn't just "buy" whenever you feel like it. You need to check if the market is open, log the time of the trade, and make sure you aren't exceeding your risk limit.

The Decorator Solution: Instead of writing "Check Risk" and "Log Time" inside every single trade function, you create a decorator called @compliance_check.

📊 The Real-World Analogy:
Imagine your boss tells you: "Every time you send me a report, you must first run it through a spell-checker and then CC the compliance team."

The Report is your Function.

The Spell-check/CC is the Decorator.

You don't change the "content" of your report; you just "wrap" it in those extra steps every time.

```python
def compliance_check(func):
    def wrapper(*args, **kwargs):
        print("🔍 [STEP 1] Checking SEBI compliance...") # Pre-action
        print("📊 [STEP 2] Checking Risk Limits...")
        result = func(*args, **kwargs) # Doing the actual task
        print("📧 [STEP 3] CC'ing the Manager...") # Post-action
        return result
    return wrapper

@compliance_check
def buy_nifty_options(strike):
    print(f"🚀 Executing Trade: Nifty {strike} Call")

# When you run this, the compliance steps happen AUTOMATICALLY
buy_nifty_options(22500)
```

📂 2. Context Managers: The "Vault"
Think of a Context Manager as a Safe or a Vault in your office.

The Analyst Scenario:
You need to get a "Physical Share Certificate" (Data) out of a high-security vault to study it.

Open: You have to use your key to open the vault.

Work: You read the document.

Close: You MUST lock the vault before you leave the office.

The Problem: If you get a phone call and run out of the office (an "Error" in code), the vault stays open. This is a security risk (a "Memory Leak").

The Context Manager Solution: The with statement acts like a "Self-Closing Door." The second you step out of that room, the door slams shut and locks itself, whether you finished your work or not.

📊 The Real-World Analogy:
Think of using a Bloomberg Terminal. You log in (Setup), you get your data (Work), and you must log out so the company doesn't get charged for extra time (Teardown).

```python
class BloombergSession:
    def __enter__(self):
        print("🔑 Logging into Bloomberg Terminal...")
        return "Terminal_Access_Granted"

    def __exit__(self, exc_type, exc_val, exc_tb):
        print("🔒 Automatically Logging out to save costs.")

# This ensures you NEVER stay logged in by mistake
with BloombergSession() as session:
    print(f"📈 Pulling historical Uranium prices...") #
    # Even if this part crashes, the 'Log out' below WILL happen.

print("Done.")
```

🏛️ 1. *args (The Unlimited List)
The * stands for "Arguments." It allows a function to take any number of positional arguments.
The Logic (The "Why")
Imagine you are writing a function to calculate the total value of a portfolio. Today the portfolio has 3 stocks. Tomorrow it might have 50.
Without *args: You would have to define 50 different variables (stock1, stock2, etc.).
With *args: You just tell Python: "Take whatever list of numbers I give you and treat them as one group."
📊 The Real-World Analogy:
Think of a Grocery Bag. You don't know exactly how many items will be in the bag, but you know you can put "Arguments" (Apples, Milk, Bread) inside it until it's full.

```python
def calculate_total_investment(*args):
    # 'args' becomes a tuple (a list that can't be changed)
    total = sum(args)
    print(f"Total Portfolio Value: ${total:,.2f}")

# You can pass 2 items...
calculate_total_investment(100, 200)

# ...or 5 items! The function doesn't care.
calculate_total_investment(100, 200, 500, 1200, 300)
```

🏗️ 2. **kwargs (The Keyword Dictionary)
The ** stands for "Keyword Arguments." It allows you to pass a dictionary (Key-Value pairs) into a function.

The Logic (The "Why")
In finance, many assets have different "Extra Details."

A Stock might have a sector.

A Bond might have a rating.

An Option might have an expiry.

If you are building a "Trade Logger," you don't want to create a new function for every asset type. You use **kwargs to capture all those extra "Labels."

📊 The Real-World Analogy:
Think of a Tax Form or a KYC Form. It has specific "Keys" (Name, PAN, Address) and "Values" (Shubh, ABCDE1234F, Goa). **kwargs is the system that reads that labeled form.

```python
def log_asset_details(ticker, **kwargs):
    print(f"Asset: {ticker}")

    # kwargs acts like a dictionary
    for key, value in kwargs.items():
        print(f"  - {key.capitalize()}: {value}")

# We can pass different labels for different assets
log_asset_details("AAPL", sector="Tech", cap="Mega")
log_asset_details("GS2030", coupon=7.5, rating="AAA")
```

```python
import functools
from datetime import datetime

# 1. THE DECORATOR (The 'Compliance' Wrapper)
def universal_trade_logger(func):
    """
    This decorator uses *args and **kwargs to log ANY trade type
    without needing to know the specific parameters beforehand.
    """
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        timestamp = datetime.now().strftime("%H:%M:%S")
        print(f"🔔 [AUDIT {timestamp}] Initializing {func.__name__.upper()}...")

        # Log positional arguments (*args) - usually the main values
        if args:
            print(f"   > Positional Data: {args}")

        # Log keyword arguments (**kwargs) - usually the extra labels
        if kwargs:
            print(f"   > Specific Details: {kwargs}")

        # Execute the actual trade
        result = func(*args, **kwargs)

        print(f"✅ [SYSTEM] Execution Successful.\n")
        return result
    return wrapper

# 2. APPLYING TO DIFFERENT FUNCTIONS
@universal_trade_logger
def execute_equity_trade(ticker, qty):
    """Simple stock trade"""
    return f"Bought {qty} of {ticker}"

@universal_trade_logger
def execute_option_trade(ticker, **option_details):
    """Complex option trade with many labels"""
    return f"Option trade placed for {ticker}"

# --- EXECUTION ---

# Case 1: Simple Stock (Uses positional arguments)
execute_equity_trade("RELIANCE", 100)

# Case 2: Complex Option (Uses **kwargs for the 'Details')
execute_option_trade("NIFTY", strike=22500, expiry="28-MAR", type="CE")
```

Section 1.5: List Comprehensions & Functional Tools
The Problem: "The Loop Overload"
As an analyst, you often need to transform data—for example, taking a list of stock prices and calculating the log return for each.

The Amateur Way: Create an empty list, write a 4-line for loop, and append each result.

The Inefficiency: In Python, standard for loops are slow. For 1 million data points, this "looping" becomes a bottleneck.

The Solution: List Comprehensions & Functional Mapping
Instead of telling Python how to loop (Step 1: start at 0, Step 2: add 1...), we tell Python what we want.

List Comprehensions: A concise, one-line way to create a new list based on an existing one.

Map & Filter: Functional tools that apply a transformation to every item in a dataset simultaneously.

📊 The Logic: The "Bulk Processing" Analogy
Think of a Post Office.

For Loop: The worker takes one letter, stamps it, walks to the bin, drops it, and walks back. (Slow).

List Comprehension/Map: The worker puts 100 letters on a tray and stamps them all in one quick motion. (Fast).

🏛️ Topic 1: List Comprehensions with Logic
You can include if/else statements directly inside the comprehension. This is extremely useful for filtering out "noise" in market data.


```python
# Simulated price changes for a portfolio of stocks
price_changes = [0.02, -0.01, 0.05, -0.04, 0.01, 0.08, -0.02]

# 1. THE OLD ANALYST WAY (For Loop)
signals = []
for change in price_changes:
    if change > 0.03:
        signals.append("BUY")
    else:
        signals.append("HOLD")

# 2. THE TYPE 2 WAY (List Comprehension)
# Syntax: [result_if_true if condition else result_if_false for item in list]
quick_signals = ["BUY" if c > 0.03 else "HOLD" for c in price_changes]

print(f"Original Changes: {price_changes}")
print(f"Processed Signals: {quick_signals}")
```

🏛️ Topic 2: Lambda Functions — The "Disposable" Logic
In ML, we often need a small function for just a second (like converting currency). We don't want to define a full def my_function(): block for a simple calculation. We use Lambda functions.

```python
# Sample prices in USD
usd_prices = [150.0, 2800.0, 450.0, 95.0]
exchange_rate = 83.0

# map(function, list) applies the lambda to every price in the list
inr_prices = list(map(lambda x: x * exchange_rate, usd_prices))

print(f"USD: {usd_prices}")
print(f"INR: {inr_prices}")
```

Section 1.6: Type Hints — Professional ML Code
The Problem: The "Dynamic Mystery"
Python is "dynamically typed," meaning it doesn't force you to say what kind of data a variable holds.

The Risk: You write a function that expects a price (number), but later in your code, you accidentally pass a ticker (string).

The Result: The code crashes in the middle of a live trade, or worse, it gives a mathematically wrong result that you don't notice until it's too late.

The Solution: Type Hinting
Type Hints allow you to explicitly state: "This variable must be an integer" or "This function returns a list of floats."

Readability: Other engineers (or your future self) know exactly what data to provide.

Bug Prevention: Modern code editors (like VS Code or Colab) will highlight errors in red before you even run the code.

Better Auto-complete: The computer can give you better suggestions because it knows the data type.

📊 The Logic: The "Currency Label" Analogy
Think of Type Hints like Currency Labels on a spreadsheet.

No Hints: A column just says 500. Is it USD? INR? Percent? You have to guess.

Type Hints: The column is clearly labeled Amount (INR). There is no room for error.

```python
# We import special types from the 'typing' module
from typing import List, Dict, Optional

# 1. VARIABLE HINTING
ticker: str = "AAPL"
price: float = 185.50
is_open: bool = True

# 2. FUNCTION HINTING
# This says: 'prices' must be a List of floats, and the function returns one float.
def calculate_average_price(prices: List[float]) -> float:
    return sum(prices) / len(prices)

# 3. ADVANCED HINTING (Optional & Union)
# Optional means it can be a float OR it can be None
def get_stop_loss(price: float, percentage: Optional[float] = None) -> float:
    if percentage:
        return price * (1 - percentage)
    return price * 0.95 # Default 5%

# --- EXECUTION ---
my_prices = [100.5, 102.0, 98.75]
avg = calculate_average_price(my_prices)

print(f"Average for {ticker}: {avg:.2f}")
```
